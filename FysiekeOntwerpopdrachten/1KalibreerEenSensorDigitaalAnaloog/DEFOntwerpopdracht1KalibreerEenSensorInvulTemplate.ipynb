## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
bits=np.array([1,2,3,4]) #gemeten hvlheid bits van de adruino
lux=np.array([1,2,3,4]) #lichtsterkte bij meting
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
import  matplotlib.pyplot as plt
from scipy.optimize import curve_fit
%matplotlib inline

#------eerste plot van data-------##
bits_err= 0.1 #onzekerheid in de volatge meting van de adruino
lux_err= 0.1  #onzekerheid in de gemeten lichtsterkte
plt.figure()
plt.errorbar(bits,lux,xerr=bits_err,yerr=lux_err,ecolor="red",fmt="k.",label="data") #data plot
plt.xlabel("$voltage/x$ (bits)")
plt.ylabel("$lichtsterkste/y$(lux)")
plt.legend()
plt.show()

#------data_fit------#
#fitfunctie:
def f(x,a,b):
    '''lineare functie fit'''
    return a*x+b

#fitten
val,cov=curve_fit(f,bits,lux) 
a,b=val #waarden voor a en b
u_a,u_b=np.sqrt(np.diag(cov)) #onzekerheden in a en b door wortel van de diagonaal van de covariantie matrix

#--------plot met beste fit-------#
x_plot=np.linspace(min(bits),max(bits),1000) #x_array om mee te plotten met hetzelfde domein als de bit metingen
plt.figure()
plt.errorbar(bits,lux,xerr=bits_err,yerr=lux_err,ecolor="red",fmt="k.",label="data") #data plot
plt.plot(x_plot,f(x_plot,*val),"r--",label=r"y = %f $\cdot$ x + %f"%(a,b))
plt.xlabel("$voltage$ (bits)")
plt.ylabel("$lichtsterkste$(lux)")
plt.legend()
plt.show()
```
