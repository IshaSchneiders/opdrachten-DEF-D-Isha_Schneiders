```python
import numpy as np
#tafel 28
bits_1=np.array([889,877,878,867,880,882,880,878]) #gemeten hvlheid bits van de adruino
mean_bits_1=np.mean(bits_1)
lux_1=np.array([810,770,760,750,750,790,788,780]) #lichtsterkte bij meting
mean_lux_1=np.mean(lux_1)
u_bits_1=np.std(bits_1)/np.sqrt(len(bits_1))
u_lux_1=np.std(lux_1)/np.sqrt(len(lux_1))


#wc dames 
bits_3=np.array([0,8,8.5,8.5,8.5,8.5,14]) #gemeten hvlheid bits van de adruino
lux_3=np.array([0,1,2,1,1,3,3]) #lichtsterkte bij meting
mean_bits_3=np.mean(bits_3)
mean_lux_3=np.mean(lux_3)
u_bits_3=np.std(bits_3)/np.sqrt(len(bits_3))
u_lux_3=np.std(lux_3)/np.sqrt(len(lux_3))

#gang midden
bits_4=np.array([680,670,684,683.5,683,682.25,683]) #gemeten hvlheid bits van de adruino
lux_4=np.array([202,200,199,197,200,199,203]) #lichtsterkte bij meting
mean_bits_4=np.mean(bits_4)
mean_lux_4=np.mean(lux_4)
u_bits_4=np.std(bits_4)/np.sqrt(len(bits_4))
u_lux_4=np.std(lux_4)/np.sqrt(len(lux_4))

#gang hoek
bits_5=np.array([325,320,340,347,347,335,325]) #gemeten hvlheid bits van de adruino
lux_5=np.array([19,20,19,20,21,19,19]) #lichtsterkte bij meting
mean_bits_5=np.mean(bits_5)
mean_lux_5=np.mean(lux_5)
u_bits_5=np.std(bits_5)/np.sqrt(len(bits_5))
u_lux_5=np.std(lux_5)/np.sqrt(len(lux_5))

#kast einde gang
bits_6=np.array([630,638,637,629,632,635,635.5]) #gemeten hvlheid bits van de adruino
lux_6=np.array([103,103,103,102,103,103,104]) #lichtsterkte bij meting
mean_bits_6=np.mean(bits_6)
mean_lux_6=np.mean(lux_6)
u_bits_6=np.std(bits_6)/np.sqrt(len(bits_6))
u_lux_6=np.std(lux_6)/np.sqrt(len(lux_6))

#gang midden
bits_7=np.array([740,753,753,744,744,744,745]) #gemeten hvlheid bits van de adruino
lux_7=np.array([277,273,277,271,278,278,274]) #lichtsterkte bij meting
mean_bits_7=np.mean(bits_7)
mean_lux_7=np.mean(lux_7)
u_bits_7=np.std(bits_7)/np.sqrt(len(bits_7))
u_lux_7=np.std(lux_7)/np.sqrt(len(lux_7))

#stoel tafel 28
bits_8=np.array([850,843,843,838,836,836,836]) #gemeten hvlheid bits van de adruino
lux_8=np.array([580,590,610,570,580,570,560]) #lichtsterkte bij meting
mean_bits_8=np.mean(bits_8)
mean_lux_8=np.mean(lux_8)
u_bits_8=np.std(bits_8)/np.sqrt(len(bits_8))
u_lux_8=np.std(lux_8)/np.sqrt(len(lux_8))

#grond tafel 28
bits_9=np.array([805,803,802,798,802,803,802]) #gemeten hvlheid bits van de adruino
lux_9=np.array([470,477,465,480,473,480,470]) #lichtsterkte bij meting
mean_bits_9=np.mean(bits_9)
mean_lux_9=np.mean(lux_9)
u_bits_9=np.std(bits_9)/np.sqrt(len(bits_9))
u_lux_9=np.std(lux_9)/np.sqrt(len(lux_9))

bits_tot=np.array([mean_bits_1,mean_bits_3,mean_bits_4,mean_bits_5,mean_bits_6,mean_bits_7,mean_bits_8,mean_bits_9])
lux_tot=np.array([mean_lux_1,mean_lux_3,mean_lux_4,mean_lux_5,mean_lux_6,mean_lux_7,mean_lux_8,mean_lux_9])
u_bits=np.array([u_bits_1,u_bits_3,u_bits_4,u_bits_5,u_bits_6,u_bits_7,u_bits_8,u_bits_9])
u_lux=np.array([u_lux_1,u_lux_3,u_lux_4,u_lux_5,u_lux_6,u_lux_7,u_lux_8,u_lux_9])
```

```python
# Plot hier de meting en fit een lijn door de metingen.
# Maak gebruik van wat je geleerd hebt bij het practicum.
import  matplotlib.pyplot as plt
from scipy.optimize import curve_fit
%matplotlib inline

#------eerste plot van data-------##
bits_err= u_bits/100 #onzekerheid in de volatge meting van de adruino
lux_err= u_lux  #onzekerheid in de gemeten lichtsterkte
plt.figure()
plt.errorbar(bits_tot/100,(lux_tot),xerr=bits_err,yerr=lux_err,ecolor="red",ms=4,fmt="k.",label="data") #data plot
plt.xlabel(r"$voltage$ ($\frac{bits}{100}$) ")
plt.ylabel("$lichtsterkste/y$(lux)")
plt.legend()
plt.show()

#------data_fit------#
#fitfunctie:
def f(x,a,b,c):
    '''exponentiele functie om mee te fitten '''
    return a*np.exp(b*x)+c

#fitten
val,cov=curve_fit(f,bits_tot/100,(lux_tot),maxfev=10000) #gedeeld door 100 omdat de fit e^800 niet kan evalueren
a,b,c=val #waarden voor a en b
u_a,u_b,u_c=np.sqrt(np.diag(cov)) #onzekerheden in a en b door wortel van de diagonaal van de covariantie matrix
print("De fitparameters zijn bepaald op a = %.1f +/- %.1f, b = %.2f +/- %.2f en c = %.0f +/- %.0f" %(a,u_a,b,u_b,c,u_c))

#--------plot met beste fit-------#
x_plot=np.linspace(min(bits_tot)/100,max(bits_tot)/100,1000) #x_array om mee te plotten met hetzelfde domein als de bit metingen
plt.figure()
plt.errorbar(bits_tot/100,(lux_tot),xerr=bits_err,yerr=lux_err,ecolor="red",fmt="k.",label="data") #data plot
plt.plot(x_plot,f(x_plot,*val),"r--",label=r"y = %.1f $\cdot$ $e^\text{%.2fx}$ %.0f"%(a,b,c))
plt.xlabel(r"$voltage$ ($\frac{bits}{100}$) ")
plt.ylabel("$lichtsterkste$(lux)")
plt.legend()
plt.show()
```
